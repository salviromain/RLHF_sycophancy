```python
import pandas as pd
df = pd.read_csv('newLlama3.csv')
df = df.dropna(subset=['final_decision'])
```

```python
nonadv_mask = (df.index % 4 == 0) | (df.index % 4 == 3)
adv_mask = (df.index % 4 == 1) | (df.index % 4 == 2)
dfnonadv = df[nonadv_mask]
dfadv = df[adv_mask]
print("First 5 rows of dfnonadv:")
display(dfnonadv.head())
print("\nFirst 5 rows of dfadv:")
display(dfadv.head())
```

# Analyze distribution of yR choice in non advesarial and adversarial settings.

```python
count_yr_not_nan_nonadv = dfnonadv['yR'].notna().sum()
count_final_decision_is_yr_nonadv = dfnonadv[dfnonadv['yR'].notna()]['final_decision'].eq('yR').sum()

print("dfnonadv Analysis:")
print(f"  'yR' is not NaN: {count_yr_not_nan_nonadv} times")
print(f"  Among non-NaN 'yR', 'final_decision' is 'yR': {count_final_decision_is_yr_nonadv} times")

print("\n")

count_yr_not_nan_adv = dfadv['yR'].notna().sum()
count_final_decision_is_yr_adv = dfadv[dfadv['yR'].notna()]['final_decision'].eq('yR').sum()

print("dfadv Analysis:")
print(f"  'yR' is not NaN: {count_yr_not_nan_adv} times")
print(f"  Among non-NaN 'yR', 'final_decision' is 'yR': {count_final_decision_is_yr_adv} times")
```
